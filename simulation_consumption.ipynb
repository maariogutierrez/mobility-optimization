```python
import pandas as pd

viajes = pd.read_csv('viajes_finales.csv')
viajes['tsO_min'] = pd.to_datetime(viajes['tsO_min'])
viajes['tsD_min'] = pd.to_datetime(viajes['tsD_min'])
viajes.drop(index=[4402, 5648, 11903, 14196, 8636, 9296, 8504, 209, 13251, 1464, 16072, 19783], inplace=True)
stations = pd.read_csv('centroides2.csv')
distribution = pd.read_json('optimos_semanales.json')
errores = pd.read_csv('errores.csv')
capacidad = pd.read_csv('capacidad_recarga.csv')
capacidad.head(1)
```

```python
import numpy as np
import pandas as pd
import ast

start_time = f'2021-02-01 00:00:00'
end_time = f'2021-02-07 23:59:00'
minutos = pd.date_range(start=start_time, end=end_time, freq='min')

bicis_semana_1 = list(np.arange(790))

station_columns = [f'station{i}' for i in range(190)]
simulations_df = pd.DataFrame(index=minutos, columns=station_columns)

simulations_df.loc[minutos[0]] = [[r['2021-02-01'][0], r['2021-02-01'][1], {bicis_semana_1.pop(0): 576 for _ in range(int(r['2021-02-01'][1]))}, np.floor(ast.literal_eval(capacidad.iloc[int(index.split('n')[1])]['Día 1'])[0]), ast.literal_eval(capacidad.iloc[int(index.split('n')[1])]['Día 1'])[0] - np.floor(ast.literal_eval(capacidad.iloc[int(index.split('n')[1])]['Día 1'])[0])] for index,r in distribution.iterrows()]
```

```python
def carga_patines(capacidad: float, patines: dict, CTE_CARGA: float = 72/60) -> dict:
  if len(patines) <= capacidad:
    for patin, carga in patines.items():
      if carga <= 576 - CTE_CARGA:
        patines[patin] = carga + CTE_CARGA
      else:
        patines[patin] = 576
  else:
    patines_ordenados = sorted(patines.items(), key=lambda x: x[1], reverse=False)
    a_cargar = patines_ordenados[:int(capacidad)]

    for patin_id, carga_actual in a_cargar:
      if carga_actual <= 576 - CTE_CARGA:
        patines[patin_id] = carga_actual + CTE_CARGA
      else:
        patines[patin_id] = 576
  return patines
```

```python
import pandas as pd
import numpy as np
import ast

saltar = []

# Iteramos sobre los minutos
for i, minuto in enumerate(minutos[1:]):
    # 1. Copiar el estado completo del minuto anterior al actual
    simulations_df.loc[minuto] = simulations_df.loc[minutos[i]]

    # -------------------------------------------------------------------------
    # CORRECCIÓN CLAVE: ACTUALIZAR CAPACIDAD Y RESTANTES AL INICIO DE HORA
    # -------------------------------------------------------------------------
    # Solo recalculamos la capacidad de carga si entramos en una nueva hora.
    # Si no es minuto 0, la capacidad y el restante se mantienen iguales al minuto anterior (ya copiados arriba).
    if minuto.minute == 0:
        for col in simulations_df.columns:
            if 'station' in col:
                try:
                    # Recuperamos el estado que acabamos de copiar del minuto anterior
                    current_capacity, current_occupation, patines, cap_carga_actual, restante_acumulado = simulations_df.loc[minuto, col]

                    # Obtenemos el valor crudo para la NUEVA hora actual
                    station_idx = int(col.split('n')[1]) # extraer ID de stationX
                    datos_hoy = ast.literal_eval(capacidad.iloc[station_idx][f'Día {minuto.day}'])
                    valor_crudo_nueva_hora = datos_hoy[minuto.hour]

                    # AQUÍ ESTÁ LA SOLUCIÓN:
                    # Sumamos el valor de la nueva hora + el restante que venía acumulado en el dataframe
                    total_disponible = valor_crudo_nueva_hora + restante_acumulado

                    nueva_capacidad_carga = np.floor(total_disponible)
                    nuevo_restante = total_disponible - nueva_capacidad_carga

                    # Actualizamos el estado del minuto actual con los nuevos valores de carga
                    simulations_df.loc[minuto, col] = (current_capacity, current_occupation, patines, nueva_capacidad_carga, nuevo_restante)

                except Exception as e:
                    # Manejo de errores si faltan datos en la tabla 'capacidad'
                    print(f"Error actualizando capacidad en {col} a las {minuto}: {e}")
                    pass

    # -------------------------------------------------------------------------
    # PROCESAMIENTO DE VIAJES (SALIDAS)
    # -------------------------------------------------------------------------
    trips_starting_now = viajes[viajes['tsO_min'] == minuto]

    for index, row in trips_starting_now.iterrows():
        origen_cluster = int(row['clusterOrigen'])
        station_col = f'station{origen_cluster}'

        if station_col in simulations_df.columns:
            # Leemos el estado actual (que ya tiene la capacidad actualizada si era necesario)
            current_capacity, current_occupation, patines, capacidad_carga, restante_carga = simulations_df.loc[minuto, station_col]

            if current_occupation == 0:
                # No hay bicis, saltamos el viaje
                saltar.append(index)
            else:
                # Lógica para elegir la bici con más carga
                max_id = 0
                max_carga = -1
                for patin, carga in patines.items():
                    if carga > max_carga:
                        max_id = patin
                        max_carga = carga

                # Retiramos el patín
                if max_id != 0:
                    del patines[max_id]

                # Actualizamos SOLO ocupación y lista de patines.
                # NO tocamos capacidad_carga ni restante_carga (se mantienen igual)
                simulations_df.loc[minuto, station_col] = (current_capacity, max(0.0, int(current_occupation) - 1.0), patines, capacidad_carga, restante_carga)

                # Registramos datos en el viaje
                viajes.loc[index, ['id', 'carga_inicial']] = [max_id, max_carga]

    # -------------------------------------------------------------------------
    # PROCESAMIENTO DE VIAJES (LLEGADAS)
    # -------------------------------------------------------------------------
    trips_ending_now = viajes[viajes['tsD_min'] == minuto]

    for index, row in trips_ending_now.iterrows():
        if index in saltar:
            continue

        destino_cluster = int(row['clusterFinal'])
        station_col = f'station{destino_cluster}'

        if station_col in simulations_df.columns:
            # Leemos estado actual (puede haber cambiado por una salida en este mismo minuto)
            current_capacity, current_occupation, patines, capacidad_carga, restante_carga = simulations_df.loc[minuto, station_col]

            if int(current_occupation) >= int(current_capacity):
                # Estación llena (lógica de rechazo, el estado no cambia)
                pass
            else:
                # Añadimos el patín con la carga restante
                patines[row['id']] = row['carga_inicial'] - row['energy_wh']

                # Actualizamos SOLO ocupación y lista de patines
                simulations_df.loc[minuto, station_col] = (current_capacity, max(0.0, int(current_occupation) + 1.0), patines, capacidad_carga, restante_carga)


    for station_col in simulations_df.columns:
        current_capacity, current_occupation, patines, capacidad_carga, restante_carga = simulations_df.loc[minuto, station_col]
        patines_antes = patines.copy()

        # Ejecutamos la carga
        patines_cargados = carga_patines(capacidad_carga, patines)

        # Guardamos el resultado en el dataframe
        simulations_df.loc[minuto, station_col] = (current_capacity, current_occupation, patines_cargados, capacidad_carga, restante_carga)

        # --- AQUÍ ESTÁ EL PRINT DE VERIFICACIÓN ---
        # Filtros: Estación 189, con hueco de carga, con patines, y al cambio de hora
        if station_col == 'station189' and capacidad_carga > 0 and len(patines) > 0 and minuto.minute == 0:
            print(f"\n⚡ VERIFICACIÓN CARGA [{minuto}] - {station_col}")
            print(f"   |> Capacidad de carga (huecos): {capacidad_carga}")
            print(f"   |> IDs y Carga ANTES:   {sorted(patines_antes.items(), key=lambda x: x[1], reverse=True)}")
            print(f"   |> IDs y Carga DESPUÉS: {sorted(patines_cargados.items(), key=lambda x: x[1], reverse=True)}")

            # Calculamos cuánto subió cada uno para verificar que sea igual a CTE_CARGA
            cambios = []
            for pid, carga_final in patines_cargados.items():
                carga_inicial = patines_antes.get(pid, 0)
                if carga_final > carga_inicial:
                    cambios.append(f"ID {pid}: +{carga_final - carga_inicial:.2f}")

            print(f"   |> Incrementos aplicados: {cambios}")
            print("-" * 60)
    # -------------------------------------------------------------------------
    # LOGS / DEBUG
    # -------------------------------------------------------------------------
    # Ejemplo de log para verificar una estación específica
    if minuto.minute == 0: # Chequear solo cambio de hora
        print(f"DEBUG {minuto}: Station 189 -> CapCarga: {simulations_df.loc[minuto, 'station189'][3]}, Restante: {simulations_df.loc[minuto, 'station189'][4]}")

    patinetes_totales = {}
    for estado_tupla in simulations_df.loc[minuto]:
         # Aseguramos desempaquetado seguro por si acaso hay nans
         if isinstance(estado_tupla, tuple):
            capacidad_estacion, ocupacion, dict_patines, capacidad_carga, restante_carga = estado_tupla
            patinetes_totales.update(dict_patines)

    # print(f'En el minuto {minuto} hay {len(patinetes_totales.keys())} patinetes totales en el sistema.')
```
